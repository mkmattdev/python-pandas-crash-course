## 2. Zmienne

Zmienna to pudełko z etykietą: wkładasz do niego wartość, a potem używasz etykiety zamiast przepisywać wartość. Znak `=` oznacza "włóż do pudełka", a nie "równa się"; do porównywania służy podwójny znak `==`.

**Zasady nazw:** małe litery i podkreślniki (`snake_case`), po angielsku, opisowe (`offers_count`, nie `x`), bez spacji i polskich znaków, nie zaczynaj od cyfry.

````python
# Cztery pudełka z danymi jednego przetargu.
offers_count = 3                   # liczba ofert w przetargu
contract_value = 1_250_000.50      # wartość umowy w zł; podkreślnik tylko dla czytelności, Python go pomija
municipality = "Radom"             # tekst zawsze w cudzysłowie; bez niego Python szukałby zmiennej Radom
is_single_bid = False              # prawda/fałsz: czy w przetargu była tylko jedna oferta
print(offers_count, contract_value, municipality, is_single_bid)   # kwota pokaże się jako 1250000.5
````

````python
# Nadpisanie: do pudełka wkładamy nową wartość, stara znika.
offers_count = 4                   # od tej linii offers_count to 4
offers_count = offers_count + 1    # weź obecną wartość, dodaj 1, włóż z powrotem: 5
print(offers_count)
````

## 3. Typy wartości

Każda wartość ma typ, tak jak komórka w Excelu ma format "liczba" albo "tekst". Funkcja `type()` mówi, jaki to typ (funkcja = gotowe polecenie z nawiasami; w nawiasie podajesz, na czym ma zadziałać).
Myśl przewodnia całego kursu: `"1 250 000,50 zł"` to TEKST i Python nic z niego nie policzy, dopóki go nie wyczyścimy i nie zamienimy na liczbę.

````python
# type() mówi, jakiego typu jest wartość w pudełku.
# Wynik wygląda tak: <class 'int'>; liczy się słowo w cudzysłowie (int), resztę pomiń.
# Pojedynczy cudzysłów ' w wynikach znaczy to samo co podwójny " w kodzie.
print(type(offers_count))      # int = liczba całkowita
print(type(contract_value))    # float = liczba z częścią dziesiętną (po KROPCE, nie po przecinku)
print(type(municipality))      # str = tekst (od "string")
print(type(is_single_bid))     # bool = prawda/fałsz (True / False)
winner = None                  # None = brak wartości, np. nieznany wykonawca
print(type(winner))            # NoneType = brak wartości; w tabelach pandas puste pole nazywa się NaN
````

````python
# Liczby się dodają, teksty się sklejają. Cudzysłów zmienia wszystko.
print(12 + 3)                     # 15, bo to liczby
print("12" + "3")                 # "123", bo to teksty: Python je skleja, nie dodaje
value_text = "1 250 000,50 zł"    # to jest TEKST, choć wygląda jak kwota
print(type(value_text))           # str: nic z tego nie policzymy, dopóki nie zamienimy na liczbę
````

## 4. Działania na liczbach

Cztery podstawowe działania to `+ - * /`, potęga to `**`, a nawiasy działają jak w matematyce. `round(x, 2)` zaokrągla x do 2 miejsc po przecinku; druga liczba mówi, ile miejsc zostawić.

````python
# Cena za tonę (Mg) = wartość umowy podzielona przez wolumen odpadów.
waste_volume = 2_480.75                          # łączny wolumen odpadów w Mg (tonach) za całą umowę
price_per_mg = contract_value / waste_volume     # dzielenie: / (wynik ma dużo cyfr po kropce)
print(price_per_mg)
print(round(price_per_mg, 2))                    # round(x, 2) = zaokrąglij x do 2 miejsc
print(24 / 12)                                   # 24 miesiące umowy to 2.0 lata; dzielenie zawsze daje float
print((1_250_000 + 980_000) / 2)                 # nawiasy: najpierw suma dwóch umów, potem średnia: 1115000.0
growth = 1.05 ** 3                               # ** to potęga: 1.05 razy 1.05 razy 1.05
print(round(growth, 4))                          # mnożnik po 3 latach waloryzacji o 5% rocznie: 1.1576
````

## 5. Operacje na tekście

Tekst ma wbudowane operacje, które wywołuje się kropką po wartości, np. `municipality.strip()`. Tak będziemy czyścić zescrapowane (pobrane automatycznie ze strony) nazwy gmin i kwoty; w notebooku 06 (Pandas: czyszczenie danych) te same operacje zadziałają na całej kolumnie naraz.

````python
# Tak wygląda nazwa gminy zescrapowana ze strony: spacje wokół i małe litery.
municipality_raw = "  radom "
print("[" + municipality_raw + "]")                # nawiasy kwadratowe pokazują, gdzie są spacje
print(len(municipality_raw))                       # len() = długość tekstu; liczy też spacje: 8
print(municipality_raw.upper())                    # WIELKIE LITERY
print(municipality_raw.lower())                    # małe litery
print(municipality_raw.title())                    # Pierwsza Litera Każdego Słowa Wielka
print("[" + municipality_raw.strip() + "]")        # strip() usuwa spacje z przodu i z tyłu
print("[" + municipality_raw + "]")                # uwaga: municipality_raw się NIE zmienił; operacje dają nowy tekst
municipality = municipality_raw.strip().title()    # dlatego wynik wkładamy do pudełka; operacje można łączyć
print(municipality)                                # "Radom"
````

````python
# .replace(A, B) podmienia w tekście każdy fragment A na B. Tu: sprzątamy kwotę zescrapowaną z ogłoszenia.
value_text = "1 250 000,50 zł"
step_1 = value_text.replace(" ", "")      # usuń spacje: "1250000,50zł"
step_2 = step_1.replace("zł", "")         # usuń jednostkę: "1250000,50"
step_3 = step_2.replace(",", ".")         # przecinek na kropkę, bo Python zna tylko kropkę: "1250000.50"
print(step_3)
print(type(step_3))                       # uwaga: to NADAL tekst (str); liczbę zrobi z niego dopiero float()
print(value_text.replace(" ", "").replace("zł", "").replace(",", "."))   # to samo w jednej linii
````

## 6. Zamiana tekstu na liczbę

Funkcje `int()`, `float()` i `str()` zamieniają typ wartości. `float()` działa tylko wtedy, gdy w tekście są same cyfry (i ewentualnie jedna kropka): bez spacji w środku, przecinka i liter.
Tego NIE uruchamiaj: Python zatrzyma się z błędem, bo w tekście są spacje, przecinek i "zł":

```python
float("1 250 000,50 zł")
# ValueError: could not convert string to float: '1 250 000,50 zł'
```

````python
# int() = na liczbę całkowitą, float() = na liczbę z częścią dziesiętną (z kropką), str() = na tekst.
offers_text = "3"                        # liczba ofert wczytana z pliku jako tekst
print(int(offers_text) + 1)              # 4: po int() to już liczba, więc dodawanie działa
print(float("1250000.50") * 2)           # tekst z KROPKĄ i bez spacji zamienia się bez problemu
print(str(2024) + "/BZP")                # liczba zamieniona na tekst, żeby skleić ją z innym tekstem
# Dwa etapy: najpierw replace() sprząta tekst, potem float() robi z niego liczbę.
value_text = "1 250 000,50 zł"
clean_text = value_text.replace(" ", "").replace("zł", "").replace(",", ".")   # "1250000.50", nadal tekst
contract_value = float(clean_text)                                              # 1250000.5, już liczba
print(contract_value / waste_volume)     # teraz działania działają: cena za Mg
````

## 7. f-string: ładne wypisywanie wyników

f-string to tekst z literą `f` przed cudzysłowem, w którym `{nawiasy klamrowe}` wstawiają wartość zmiennej; po dwukropku podajesz format, np. `:.2f` = 2 miejsca po przecinku.
Python formatuje liczby po angielsku: przecinek oddziela tysiące, kropka części dziesiętne.

````python
# f-string: w nawiasach klamrowych wstawiamy zmienną, po dwukropku podajemy format.
print(f"Gmina: {municipality}, liczba ofert: {offers_count}")
print(f"Cena za Mg: {price_per_mg:.2f} zł")               # .2f = 2 miejsca po przecinku
print(f"Wartość umowy: {contract_value:,.2f} zł")         # , = separator tysięcy (angielski przecinek)
single_bid_share = 20 / 49                                 # 2019: 20 z 49 przetargów (ze znaną liczbą ofert) miało jedną ofertę
print(f"Udział przetargów z 1 ofertą: {single_bid_share:.1%}")   # .1% = procent z 1 miejscem po przecinku
````

**Podsumowanie**
- Zmienna to pudełko z etykietą; `=` wkłada do niego wartość, a nowa wartość zastępuje starą.
- Typ ma znaczenie: `12` to liczba, `"12"` to tekst; `type()` mówi, co masz w pudełku.
- Tekst sprzątasz przez `.strip()`, `.title()`, `.replace()`, a na liczbę zamieniasz przez `float()`; w notebooku 06 (Pandas: czyszczenie danych) zrobimy to samo dla całej kolumny naraz.
- f-string wypisuje wyniki: `{x:.2f}`, `{x:,.2f}`, `{x:.1%}`.
