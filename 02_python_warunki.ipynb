**Czego się nauczysz**
- Porównania (`==`, `<`, `>` ...) dają odpowiedź `True` albo `False`, jak filtr w Excelu.
- Łączenie warunków: `and`, `or`, `not`.
- `in`: czy tekst zawiera fragment (np. kod odpadów).
- `if / elif / else`: Python podejmuje decyzję; dwukropek i wcięcia.
- Sprawdzenie, czy w polu jest liczba, czy `brak danych`, zanim coś policzysz.

## 1. Porównania: pytanie, na które odpowiedź to True albo False

Porównanie to pytanie do Pythona: "czy liczba ofert równa się 1?". Odpowiedź jest zawsze `True` (prawda) albo `False` (fałsz).
To dokładnie to, co robi filtr w Excelu ("pokaż wiersze, gdzie Liczba ofert = 1"); w pandas ten sam warunek nałożymy na całą kolumnę.

````python
# Dane jednego przetargu w dwóch zmiennych (pudełkach z etykietą).
offers_count = 1                  # liczba ofert
contract_value = 1_250_000.50     # wartość umowy w zł

# Każde porównanie daje True albo False.
print(offers_count == 1)                 # == czy równe? (UWAGA: dwa znaki; jeden znak = wkłada wartość do zmiennej)
print(offers_count != 1)                 # != czy różne?
print(contract_value > 1_000_000)        # >  czy większe?
print(contract_value < 1_000_000)        # <  czy mniejsze?
print(contract_value >= 1_250_000.50)    # >= czy większe lub równe?
print(contract_value <= 1_000_000)       # <= czy mniejsze lub równe?
````

````python
# Wynik porównania można zapisać do zmiennej: to typ bool (True albo False).
is_single_bid = offers_count == 1
print(is_single_bid)

# Pułapka: tekst "1" (w cudzysłowie) to NIE liczba 1. Tak wygląda kolumna "Liczba ofert" po scrapowaniu (automatycznym pobraniu ze strony).
print("1" == 1)          # False, bo tekst i liczba to różne typy
print(int("1") == 1)     # True: int() najpierw zamienia tekst "1" na liczbę 1, potem porównujemy
````

## 2. Łączenie warunków: and, or, not

`and` = oba warunki muszą być prawdziwe, `or` = wystarczy jeden, `not` = odwraca odpowiedź.
Przykład: przetarg "wart uwagi" to taki, w którym była jedna oferta I wartość przekracza 1 mln zł.
⚠️ W pandas, gdy warunek dotyczy całej kolumny, zamiast `and`/`or` piszemy `&`/`|`, a każdy warunek w nawiasach; tak zrobimy w notebooku 05 (Pandas: start).

````python
offers_count = 1
contract_value = 1_250_000.50
contract_months = 24              # okres umowy w miesiącach

# and: jedna oferta I duża wartość (oba warunki muszą być prawdą).
print(offers_count == 1 and contract_value > 1_000_000)

# or: wystarczy, że jeden z warunków jest prawdą (tu: pierwszy tak, drugi nie, więc True).
print(offers_count == 1 or contract_months >= 36)

# not: odwraca odpowiedź (True -> False, False -> True); nawias pokazuje, co dokładnie odwracamy.
print(not (offers_count == 1))
````

**Twoja kolej:** dokończ warunek: przetarg "wart uwagi" to jedna oferta LUB wartość powyżej 1 mln zł. Tu są dwie oferty, więc o wyniku zdecyduje wartość.

````python
offers_count = 2                  # tym razem dwie oferty
contract_value = 1_250_000.50     # wartość umowy w zł
# TODO: zamień False na porównanie wartości umowy (contract_value) z liczbą 1_000_000.
print(offers_count == 1 or False)
````

## 3. Czy tekst zawiera fragment: in

`in` pyta: "czy ten fragment jest w tym tekście?". Przyda się do kodów odpadów i do sprawdzenia, w jakim formacie przyszła kwota.

````python
waste_codes = "20 03 01; 15 01 01; 20 01 08"    # kody odpadów z jednego przetargu (jeden tekst)
value_text = "1 250 000,50 zł"                   # wartość umowy, jeszcze jako tekst

print("20 03 01" in waste_codes)    # czy przetarg obejmuje odpady zmieszane (kod 20 03 01)?
print("17 01 07" in waste_codes)    # tego kodu nie ma, więc False
print("zł" in value_text)           # czy kwota ma dopisek "zł"?
print("PLN" in value_text)          # a może "PLN"?
print("zł" not in value_text)       # not in = czy NIE zawiera

# Pułapka: wielkość liter ma znaczenie, "mg" to nie "Mg".
print("mg" in "12 500,00 Mg")
````

## 4. Decyzja: if / elif / else

`if` = "jeżeli warunek jest prawdziwy, wykonaj to, co pod spodem". Po warunku stawiamy dwukropek, a kod do wykonania wcinamy o 4 spacje: po wcięciu Python poznaje, co należy do `if`.
`elif` = "a jeśli nie, sprawdź kolejny warunek", `else` = "w każdym innym przypadku".

````python
offers_count = 1

# if ... else: dwie możliwości.
if offers_count == 1:                             # dwukropek na końcu linii!
    print("Jedna oferta: brak konkurencji")       # 4 spacje wcięcia = ta linia należy do if
else:
    print("Więcej ofert: jest konkurencja")

print("Ta linia nie jest wcięta, więc wykona się zawsze.")
````

````python
offers_count = 2

# Trzy poziomy: Python sprawdza warunki po kolei, wykonuje PIERWSZY prawdziwy, resztę pomija.
if offers_count == 1:
    competition_level = "brak konkurencji"
elif offers_count == 2:
    competition_level = "słaba konkurencja"
else:                                   # czyli 3 oferty i więcej
    competition_level = "konkurencja"

# f-string: litera f przed cudzysłowem, a w klamrach {} Python wstawia wartość zmiennej.
print(f"Liczba ofert: {offers_count}, ocena: {competition_level}")
# W notebooku 04 (Funkcje) zamienimy ten if w funkcję competition_level(offers_count), żeby nie pisać go od nowa dla każdego przetargu.
````

**Częste błędy** (przeczytaj, nie uruchamiaj):
```python
if offers_count = 1:      # co by się stało: SyntaxError, bo jeden znak = wkłada wartość, a == porównuje
if offers_count == 1      # co by się stało: SyntaxError, brakuje dwukropka na końcu
if offers_count == 1:
print("jedna oferta")     # co by się stało: IndentationError, brakuje wcięcia (4 spacje)
```

**Twoja kolej:** w komórce wyżej (tej z `elif`) zmień `offers_count` na 3, potem na 1, i za każdym razem uruchom ją ponownie. Potem popraw warunek w `if` w komórce niżej.

````python
contract_value = 3_000_000.00     # sprawdź też dla 12_500_000.00

# TODO: popraw warunek tak, żeby "duży przetarg" wypisało się tylko, gdy wartość przekracza 10 mln zł.
if contract_value > 0:
    print("duży przetarg")
else:
    print("zwykły przetarg")
````

## 5. Brakujące dane: sprawdź, zanim policzysz (opcjonalnie, jeśli zostanie czas)

W scrapowanej kolumnie "Liczba ofert" zamiast liczby bywa tekst `brak danych` (albo puste pole), a `int("brak danych")` skończyłoby się błędem `ValueError` (Python nie umie zrobić liczby z takiego tekstu).
Dlatego najpierw pytamy `if`, co przyszło. W pandas zrobi to za nas jedno polecenie, ale zasada jest ta sama: sprawdź, zanim policzysz.

````python
offers_text = "brak danych"     # tak wygląda brak w kolumnie "Liczba ofert"; zmień na "2" i uruchom ponownie
# Uwaga: puste pole "" ten if przepuści i int("") da ten sam błąd; jak je wyłapać (.isdigit()), pokazuje zadanie 4 (bonus) na końcu tego notebooka.

if offers_text == "brak danych":
    offers_count = None                # None = brak wartości ("nic tu nie ma"), zamiast liczby
else:
    offers_count = int(offers_text)    # normalny przypadek: tekst "2" -> liczba 2

print(offers_count)
````

**Podsumowanie**
- Porównania `==`, `!=`, `<`, `>`, `<=`, `>=` dają `True` albo `False`; pamiętaj, że tekst `"1"` to nie liczba `1`.
- `and` (oba warunki), `or` (wystarczy jeden), `not` (odwraca) łączą warunki; w pandas ich odpowiedniki to `&` i `|` z nawiasami.
- `in` sprawdza, czy tekst zawiera fragment, np. kod `20 03 01` albo dopisek `zł`.
- `if / elif / else`: dwukropek, wcięcie 4 spacje, `==` do porównania; Python wykonuje pierwszy prawdziwy warunek.
- Zanim zamienisz tekst na liczbę, sprawdź `if`-em, czy to nie `brak danych`; `None` = brak wartości.
