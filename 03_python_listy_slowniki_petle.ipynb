## 1. Lista: kolumna wartości

Lista to wiele wartości w jednej zmiennej, zapisanych w nawiasach kwadratowych po przecinku.
Myśl o niej jak o kolumnie `Liczba ofert` dla kilku przetargów.
**Pułapka:** Python numeruje elementy od 0, więc pierwszy element to `[0]`.

```python
# Lista = kilka wartości w jednej zmiennej, w nawiasach kwadratowych, po przecinku.
offers = [1, 2, 1, 3, 1]                  # liczba ofert w pięciu przetargach (jak kolumna w Excelu)
print(offers)
print("Ile przetargów:", len(offers))     # len = długość listy, czyli liczba elementów
print("Pierwszy przetarg:", offers[0])    # UWAGA: numeracja od 0, więc [0] to pierwszy element
print("Ostatni przetarg:", offers[-1])    # -1 = ostatni element (liczone od końca)
# offers[5] dałoby błąd IndexError (index = numer elementu): lista ma 5 elementów o numerach 0, 1, 2, 3, 4
```

```python
# Dopisywanie i szybkie podsumowania listy (jak SUMA, MAX, MIN w Excelu).
offers = [1, 2, 1, 3, 1]                  # ta sama lista co w poprzedniej komórce, celowo wpisana od nowa (dlaczego: pułapka niżej)
offers.append(2)                          # append = dopisz nowy element na koniec listy
# Pułapka: append zmienia listę na stałe; bez linii "offers = [...]" na górze komórki każde jej ponowne uruchomienie dopisałoby kolejne 2
print("Po dopisaniu:", offers)
print("Suma ofert:", sum(offers))         # sum = dodaje wszystkie elementy
print("Najwięcej ofert:", max(offers))    # max = największy element
print("Najmniej ofert:", min(offers))     # min = najmniejszy element
print("Posortowane:", sorted(offers))     # sorted = nowa, posortowana lista (oryginał bez zmian)
average_offers = sum(offers) / len(offers)    # średnia = suma podzielona przez liczbę elementów
print(f"Średnia liczba ofert: {average_offers:.2f}")   # :.2f = 2 miejsca po przecinku
```

**Twoja kolej:** uzupełnij listę okresów umów i policz średni okres.

```python
# TODO: dopisz do listy okresy umów jeszcze dwóch przetargów (np. 12 i 48) i uruchom komórkę ponownie
# Podpowiedź: średnia = sum(...) / len(...)
contract_months_list = [24, 36]           # <- dopisz tu dwie liczby po przecinku
average_months = sum(contract_months_list) / len(contract_months_list)   # średni okres umowy
print(f"Średni okres umowy: {average_months:.1f} miesięcy")   # :.1f = 1 miejsce po przecinku
```

## 2. Pętla for: dla każdego elementu zrób...

Pętla `for` bierze elementy listy po kolei i dla każdego wykonuje wcięty blok kodu.
Tak jak w `if`: dwukropek na końcu linii i 4 spacje wcięcia.

⚠️ W pandas zwykle NIE piszemy pętli, bo pandas robi to samo dla całej kolumny naraz; ale trzeba rozumieć, co dzieje się w środku.

```python
# Pętla for: "dla każdej liczby ofert z listy offers wykonaj wcięty blok".
for offers_count in offers:               # offers_count = kolejny element listy (przy każdym przejściu pętli inny)
    print("Liczba ofert w przetargu:", offers_count)   # wcięcie 4 spacje = to jest "w środku" pętli
print("Koniec pętli")                     # bez wcięcia = wykona się raz, po zakończeniu pętli
# To samo z if w środku: licznik przetargów, które miały tylko jedną ofertę.
single_bid_count = 0                                  # licznik startuje od zera
for offers_count in offers:                           # dla każdego przetargu...
    if offers_count == 1:                             # ...jeśli była jedna oferta (== porównanie, nie =)
        single_bid_count = single_bid_count + 1       # ...dolicz 1 (podwójne wcięcie: if jest w środku for)
single_bid_share = single_bid_count / len(offers)     # udział = ile z jedną ofertą / ile wszystkich
print(f"Przetargi z jedną ofertą: {single_bid_count} z {len(offers)}, czyli {single_bid_share:.1%}")   # :.1% = ułamek jako procent, 1 miejsce po przecinku
```

**Twoja kolej:** zmień warunek tak, żeby policzyć przetargi z co najmniej 2 ofertami.

```python
# TODO: zmień warunek w if tak, żeby licznik zliczał przetargi z co najmniej 2 ofertami
# Podpowiedź: >= znaczy "większe lub równe"
competitive_count = 0                     # licznik startuje od zera
for offers_count in offers:               # dla każdego przetargu...
    if offers_count == 3:                 # <- TODO: tu zmień warunek
        competitive_count = competitive_count + 1   # ...dolicz 1
print("Przetargi spełniające warunek:", competitive_count)
```

## 3. Pętla po tekście

Tekst też jest ciągiem elementów: pętla `for` przechodzi po nim znak po znaku.
Dla każdego znaku (w kodzie nazwiemy go `char`) `char.isdigit()` odpowiada `True`/`False` na pytanie "czy ten znak to cyfra?".
W notebooku 04 (Funkcje) zbudujemy z tej pętli funkcję `clean_amount`, która wyciąga liczbę z `"1 250 000,50 zł"`.

```python
# Pętla po tekście: sprawdzamy każdy znak kwoty i zbieramy same cyfry.
value_text = "1 250,50"                   # kwota jako TEKST (tak wygląda kwota pobrana ze strony z ogłoszeniem)
digits = ""                               # tu zbieramy cyfry; na start pusty tekst
for char in value_text:                   # char = kolejny znak tekstu (cyfra, spacja albo przecinek)
    print("znak:", char, "| cyfra?", char.isdigit())   # spacja to też znak, choć jej nie widać
    if char.isdigit():                    # isdigit = True tylko dla cyfr 0-9
        digits = digits + char            # doklejamy cyfrę na koniec zbieranego tekstu
print("Same cyfry:", digits)              # spacja i przecinek odpadły
# To jeszcze NIE jest kwota 1250,50: przecinek zachowa dopiero funkcja clean_amount, którą zbudujemy z tej pętli w notebooku 04 (Funkcje)
```

## 4. Słownik: etykieta → wartość

Słownik to zbiór par `"klucz": wartość` w nawiasach klamrowych.
To jak jeden wiersz tabeli z nagłówkami: klucz (nagłówek) mówi, co oznacza dana wartość.
⚠️ Pułapka: klucz podajemy w cudzysłowie, bo to tekst: `tender["gmina"]`, nie `tender[gmina]`.

```python
# Słownik = wartości z etykietami (kluczami), zapis: {"klucz": wartość, ...}.
tender = {"gmina": "Radom", "liczba_ofert": 2, "wartosc_pln": 1_250_000.50}   # jeden przetarg = jeden wiersz; podkreślniki w liczbie tylko dla czytelności
print(tender)                             # w wydruku cudzysłowy są pojedyncze ('gmina'); dla Pythona to to samo co "gmina" w kodzie
print("Gmina:", tender["gmina"])          # odczyt po kluczu (klucz w cudzysłowie!)
print("Wartość:", tender["wartosc_pln"], "zł")
tender["wolumen_mg"] = 2_500.0            # nowy klucz = nowa "kolumna" w tym wierszu
print("Po dodaniu wolumenu:", tender)
# tender["okres"] dałoby błąd KeyError: takiego klucza w słowniku nie ma
```

**Bezpieczny odczyt `.get()` i pętla po parach `.items()` (opcjonalnie, jeśli zostanie czas)**

`.get()` przyda się w zadaniu 5 (bonus); jeśli pomijamy tę komórkę na zajęciach, uruchom ją w domu.

```python
# .get() = bezpieczny odczyt: gdy klucza nie ma, zwraca wartość zapasową zamiast błędu.
print("Wykonawca:", tender.get("wykonawca", "brak danych"))   # klucza nie ma -> "brak danych"
print("Gmina:", tender.get("gmina", "brak danych"))           # klucz jest -> zwykła wartość
# Pętla po parach klucz -> wartość (.items() = "daj mi wszystkie pary").
for key, value in tender.items():         # key = etykieta, value = wartość; przy każdym przejściu pętli inna para
    print(f"{key}: {value}")
```

**Słownik jako mapa: stara nazwa → nowa nazwa (opcjonalnie, jeśli zostanie czas)**

Taki słownik to gotowa mapa do zmiany nagłówków kolumn z Excela (`"Liczba ofert"`) na nazwy wygodne w kodzie (`liczba_ofert`): bez spacji i polskich znaków.
Jeśli pomijamy tę komórkę na zajęciach, uruchom ją w domu.

```python
# Słownik jako mapa: nazwa kolumny z Excela -> nazwa wygodna w kodzie (bez spacji i polskich znaków).
COLUMN_NAMES = {                          # WIELKIE LITERY to zwyczaj: ta zmienna to "stała", nie zmieniamy jej
    "Gmina": "gmina",
    "Liczba ofert": "liczba_ofert",
    "Wartość umowy": "wartosc_umowy",
    "Wolumen odpadów": "wolumen_odpadow",
}
print(COLUMN_NAMES["Liczba ofert"])       # pytamy mapę: jak ma się nazywać kolumna "Liczba ofert"?
```

## 5. Lista słowników = tabela

Kilka słowników w jednej liście to cała tabela: każdy słownik to wiersz, a klucze to nagłówki kolumn.
Pętla po takiej liście robi to samo dla każdego wiersza, jak formuła przeciągnięta w dół w Excelu.

```python
# Lista słowników = tabela: 4 przetargi (wiersze), każdy z tymi samymi kluczami (kolumnami).
tenders = [
    {"gmina": "Radom", "liczba_ofert": 2, "wartosc_pln": 1_250_000.50, "wolumen_mg": 2_500.0},
    {"gmina": "Płock", "liczba_ofert": 1, "wartosc_pln": 3_600_000.00, "wolumen_mg": 4_000.0},
    {"gmina": "Siedlce", "liczba_ofert": 3, "wartosc_pln": 890_000.00, "wolumen_mg": 1_850.0},
    {"gmina": "Ciechanów", "liczba_ofert": 1, "wartosc_pln": 2_150_000.00, "wolumen_mg": 2_300.0},
]
print("Liczba przetargów w tabeli:", len(tenders))   # len na liście = liczba wierszy
print("Drugi wiersz:", tenders[1])                    # [1] = drugi element (numeracja od 0)
```

```python
# Pętla po wierszach: dla każdego przetargu liczymy cenę za Mg i dopisujemy ją jako nową "kolumnę".
# Uwaga: pętla nadpisuje zmienną tender; po pętli tender to ostatni wiersz (Ciechanów), a nie słownik z Radomiem.
for tender in tenders:                                            # tender = kolejny wiersz (słownik)
    price_per_mg = tender["wartosc_pln"] / tender["wolumen_mg"]   # cena za tonę = wartość / wolumen
    tender["cena_za_mg"] = round(price_per_mg, 2)                 # nowy klucz w tym wierszu
    print(f"{tender['gmina']}: {tender['cena_za_mg']:.2f} zł/Mg")   # klucz w 'pojedynczych' cudzysłowach (dlaczego: pułapka niżej)
# Pułapka: w f"..." klucz piszemy w 'pojedynczych' cudzysłowach: w Pythonie starszym niż 3.12 podwójny cudzysłów w środku
# zakończyłby tekst za wcześnie i dał błąd; w nowszym działa, ale pojedyncze są bezpieczne w każdej wersji
```

**Zapowiedź pandas:** taką listę słowników pandas zamienia jednym poleceniem w DataFrame,
czyli tabelę jak arkusz w Excelu (tę komórkę tylko uruchom i obejrzyj wynik, nic nie dopisujesz).
Od notebooka 05 (Pandas: start) pracujemy już tylko na takich tabelach.

```python
import pandas as pd                       # pandas = biblioteka (gotowy dodatek do Pythona) do pracy z tabelami; "pd" to przyjęty skrót
pd.DataFrame(tenders)                     # DataFrame, czyli tabela: klucze stały się nagłówkami kolumn
```
